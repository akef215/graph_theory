```python
from heap import Heap
from graphe import Graph
import numpy as np
```

```python
edges = [(0, 1, 3), (1, 2, 7), (2, 3, 6), (3, 4, 8), (4, 7, 9), (7, 8, 1), (5, 6, 0), \
        (0, 5, 7), (1, 5, 10), (2, 8, 1), (8, 5, 16), (8, 6, 5), (2, 7, 3), (3, 7, 13)]

G = Graph(oriented=False, valued=True)
G.load_from_weighted_edges(edges)
H = Heap(type='min')
np.set_printoptions(precision=2)
origin = 0
```

```python
# Initialise the distances and predecessors arrays
d = np.ones(G.size_) * np.inf
d[origin] = 0
pi = -np.ones(G.size_)

# Initialise the 'min' heap
tas = zip(G.vertices_, d)
H.init_heap(tas)
H.show()
```

```python
%%time
"""
    Une implementation efficace de l'algorithme de Prim
    pour trouver un arbre de poids minimum d'un graphe G
    en utilisant un tas de type min, où la priorité dans
    ce tas est défini par le poids d'arrete qui relie le
    sommet en question avec l'ACM, np.inf sinon

    Abréviation:
        ACM : Arbre Couvrant de poids Minimum
"""

verbose = False
ACM = []
while not H.is_empty():
    # récupérer le sommet le plus proche de l'ACM
    curr_vertex, curr_priority = H.dequeue()

    # les successeurs du curr_vertex
    succ_curr_vertex = G.next(curr_vertex)
    if verbose:
        print(f"current vertex :\n{curr_vertex} -> {succ_curr_vertex}")

    for vertex in succ_curr_vertex:
        # On vérifie si le sommet est dans le tas
        # et si on peut améliorer son estimation  
        if vertex in H.positions_ and d[vertex] > G.weight(curr_vertex, vertex):
            # On met a jour la priorité du vertex
            d[vertex] = G.weight(curr_vertex, vertex)
            pi[vertex] = curr_vertex

            # On ajoute l'arc curr_vertex -> vertex a l'ACM
            ACM.append((curr_vertex, vertex, d[vertex]))

            # On met a jour le tas
            H.update_priority(vertex, d[vertex])
            if verbose:
                H.show()
```

```python
print("l'arbre couvrant de poids minimum :", pi)
children = dict()
length = 0
for i, v in enumerate(pi):
    if v == -1:
        continue
    children.setdefault(v, []).append(i)
    length += G.weight(v, i)

print(f"poids de l'arbre {length}")
```

```python
from bigtree import Node

def build_tree(root, tree):
    parent = Node(str(root + 1))
    if tree.get(root) is None:
        return parent
    for c in tree[root]:
        child = build_tree(c, tree)
        child.parent = parent  
    return parent
```

```python
build_tree(0, children).vshow()
print(f"poids de l'arbre {length}")
```
