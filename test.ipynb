```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("test.csv")
df.head()
```

```python
from datetime import datetime
t = df.iloc[1]["Temps"].split(";")
t.append('18:45:45')
if len(t) % 2 == 1:
    t.append(t[-1]) 

out = []
for ti in t:
    format = '%H:%M:%S'
    ti_clean = ti.strip()
    #print(ti_clean)
    #print(datetime.strptime(ti_clean, format))
    out.append(datetime.strptime(ti_clean, format))

day = 0
for i in range(0, len(out), 2):
    time1 = datetime.strptime("6:20:50", "%H:%M:%S")
print('Start time is :', time1.time())

time2 = datetime.strptime("11:56:18", "%H:%M:%S")
print('End time is :', time2.time())


delta = time2 - time1


print("Time difference in seconds is", delta.total_seconds(), "seconds")

print(day)    
    
```
